```python
from transformers import CLIPProcessor, CLIPModel

model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")

print("AI model loaded successfully!")
```

```python
from PIL import Image

image1 = Image.open("/content/bag1.jpg.jpeg").convert("RGB")
image2 = Image.open("/content/bag2.jpg.jpeg").convert("RGB")

print("Image 1 loaded:", image1.size)
print("Image 2 loaded:", image2.size)
```

```python
import torch

with torch.no_grad():
    output1 = model.vision_model(
        **processor(images=image1, return_tensors="pt")
    )

    output2 = model.vision_model(
        **processor(images=image2, return_tensors="pt")
    )

    image_features1 = model.visual_projection(output1.pooler_output)
    image_features2 = model.visual_projection(output2.pooler_output)

print("Image 1 features:", image_features1.shape)
print("Image 2 features:", image_features2.shape)
```

```python
import torch.nn.functional as F

# Normalize the image features
image_features1 = F.normalize(image_features1, p=2, dim=-1)
image_features2 = F.normalize(image_features2, p=2, dim=-1)

# Calculate similarity
similarity = torch.sum(image_features1 * image_features2).item()

print("Similarity score:", similarity)
print("Similarity percentage:", round(similarity * 100, 2), "%")
```

```python
pillow_image = Image.open("/content/pillow.jpg.jpeg").convert("RGB")

print("Pillow image loaded:", pillow_image.size)
```

```python
def compare_images(image_a, image_b):
    # Convert both images into CLIP features
    with torch.no_grad():
        output_a = model.vision_model(
            **processor(images=image_a, return_tensors="pt")
        )

        output_b = model.vision_model(
            **processor(images=image_b, return_tensors="pt")
        )

        features_a = model.visual_projection(output_a.pooler_output)
        features_b = model.visual_projection(output_b.pooler_output)

    # Normalize features
    features_a = F.normalize(features_a, p=2, dim=-1)
    features_b = F.normalize(features_b, p=2, dim=-1)

    # Calculate similarity
    score = torch.sum(features_a * features_b).item()

    return score
```

```python
from sentence_transformers import SentenceTransformer

text_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Text AI model loaded successfully!")
```

```python
lost_description = "Black backpack with a blue keychain attached to the zipper."

found_description = "Black bag with a small blue keychain on the zip."

print("Lost:", lost_description)
print("Found:", found_description)
```

```python
from sentence_transformers.util import cos_sim

lost_embedding = text_model.encode(lost_description, convert_to_tensor=True)
found_embedding = text_model.encode(found_description, convert_to_tensor=True)

text_similarity = cos_sim(lost_embedding, found_embedding).item()

print("Description similarity:", round(text_similarity * 100, 2), "%")
```

```python
different_description = "Blue water bottle with a silver cap."

different_embedding = text_model.encode(
    different_description,
    convert_to_tensor=True
)

different_similarity = cos_sim(
    lost_embedding,
    different_embedding
).item()

print("Different item similarity:", round(different_similarity * 100, 2), "%")
```

```python
def compare_descriptions(description_a, description_b):
    embedding_a = text_model.encode(
        description_a,
        convert_to_tensor=True
    )

    embedding_b = text_model.encode(
        description_b,
        convert_to_tensor=True
    )

    score = cos_sim(embedding_a, embedding_b).item()

    return score
```

```python
from math import radians, sin, cos, sqrt, atan2

lost_lat = 17.3850
lost_lon = 78.4867

found_lat = 17.3860
found_lon = 78.4875

R = 6371  # Earth's radius in km

lat1 = radians(lost_lat)
lat2 = radians(found_lat)

dlat = radians(found_lat - lost_lat)
dlon = radians(found_lon - lost_lon)

a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
c = 2 * atan2(sqrt(a), sqrt(1 - a))

distance = R * c

print("Distance:", round(distance, 3), "km")
```

```python
location_score = max(0, 1 - (distance / 5))

print("Location similarity:", round(location_score * 100, 2), "%")
```

```python
from transformers import CLIPProcessor, CLIPModel

model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")

print("AI model loaded successfully!")
```

```python
image_score = compare_images(image1, image2)

description_score = compare_descriptions(
    lost_description,
    found_description
)

overall_score = (
    0.50 * image_score +
    0.30 * description_score +
    0.20 * location_score
)

print("Image Score:", round(image_score * 100, 2), "%")
print("Description Score:", round(description_score * 100, 2), "%")
print("Location Score:", round(location_score * 100, 2), "%")
print("Overall Match Score:", round(overall_score * 100, 2), "%")
```

```python
def calculate_match_score(image_score, description_score, location_score):
    overall_score = (
        0.50 * image_score +
        0.30 * description_score +
        0.20 * location_score
    )

    return overall_score
```

```python
def match_result(overall_score):
    if overall_score >= 0.75:
        return "Likely Match"
    else:
        return "Not a Match"
```

```python
result = match_result(overall_score)

print("Result:", result)
```

```python
def find_match(image_a, image_b, description_a, description_b, location_score):

    image_score = compare_images(image_a, image_b)

    description_score = compare_descriptions(
        description_a,
        description_b
    )

    overall_score = calculate_match_score(
        image_score,
        description_score,
        location_score
    )

    result = match_result(overall_score)

    print("Image Score:", round(image_score * 100, 2), "%")
    print("Description Score:", round(description_score * 100, 2), "%")
    print("Location Score:", round(location_score * 100, 2), "%")
    print("Overall Match:", round(overall_score * 100, 2), "%")
    print("Result:", result)

    return overall_score
```

```python
find_match(
    image1,
    image2,
    lost_description,
    found_description,
    location_score
)
```
